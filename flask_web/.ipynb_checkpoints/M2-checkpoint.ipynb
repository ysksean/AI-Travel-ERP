# 데이터 로드

```python
import os
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, ElectraForSequenceClassification
from torch.optim import AdamW
from tqdm.notebook import tqdm  # 주피터용 진행바

# ==========================================
# 1. 설정 (Configuration)
# ==========================================
# 테스트할 때는 1~2번만, 실제 학습할 때는 10번 이상 설정하세요.
EPOCHS = 4              # [변경] 2 -> 4 (심화 학습을 위해 조금 늘림)
LEARNING_RATE = 5e-5    
BATCH_SIZE = 16       
MAX_LEN = 128           
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 파일 경로 (현재 주피터 노트북 위치 기준)
BASE_DIR = os.getcwd()
DATA_FILE = "../ERP 필요한 데이터/6. 랜드사한테 고객 요청 답변 내용_집에서추가/3중분류_생성.csv"
MODEL_SAVE_DIR = os.path.join(BASE_DIR, "models", "koelectra_sentiment")

# 라벨 맵핑 (0:불가, 1:확정, 2:보류)
LABEL_MAP = {0: "불가(부정)", 1: "확정(긍정)", 2: "보류(중립)"}

print(f"🚀 사용 장치: {DEVICE}")

# ==========================================
# 2. 데이터 로드 및 확인
# ==========================================
if not os.path.exists(DATA_FILE):
    print(f"❌ 오류: '{DATA_FILE}' 파일이 없습니다. 업로드해주세요.")
else:
    df = pd.read_csv(DATA_FILE)
    print(f"✅ 데이터 로드 성공! 총 {len(df)}개")
    print(df.head()) # 상위 5개 출력
```

# 모델 준비

```python
# ==========================================
# 3. 데이터셋 클래스 정의
# ==========================================
class SentimentDataset(Dataset):
    def __init__(self, dataframe, tokenizer, max_len):
        self.data = dataframe
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):
        row = self.data.iloc[index]
        text = str(row['text'])
        label = int(row['label'])

        # 토크나이징 (문장 -> 숫자)
        # padding='max_length'로 길이를 128로 고정하여 에러 방지
        encoding = self.tokenizer.encode_plus(
            text,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )

        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(label, dtype=torch.long)
        }

# ==========================================
# 4. 모델 및 토크나이저 로드
# ==========================================
print("⏳ 모델 로딩 중...")
model_name = "monologg/koelectra-base-v3-discriminator"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# num_labels=3 (불가, 확정, 보류)
model = ElectraForSequenceClassification.from_pretrained(model_name, num_labels=3)
model.to(DEVICE)

# 데이터 로더 연결
dataset = SentimentDataset(df, tokenizer, MAX_LEN)
loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)
optimizer = AdamW(model.parameters(), lr=LEARNING_RATE)

print("✅ 모델 준비 완료!")
```

# 학습

```python
from tqdm import tqdm
# ==========================================
# 5. 학습 루프 (Training Loop)
# ==========================================
print(f"🔥 학습 시작 (총 {EPOCHS}회)")
model.train()

for epoch in range(EPOCHS):
    total_loss = 0
    progress_bar = tqdm(loader, desc=f"Epoch {epoch+1}/{EPOCHS}")
    
    for batch in progress_bar:
        input_ids = batch['input_ids'].to(DEVICE)
        attention_mask = batch['attention_mask'].to(DEVICE)
        labels = batch['labels'].to(DEVICE)

        optimizer.zero_grad()
        
        # 순전파 (Forward)
        outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
        
        loss = outputs.loss
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        progress_bar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(loader)
    print(f"   Epoch {epoch+1} 평균 오차(Loss): {avg_loss:.4f}")

print("\n🎉 학습 완료!")

# ==========================================
# 6. 모델 저장
# ==========================================
if not os.path.exists(MODEL_SAVE_DIR):
    os.makedirs(MODEL_SAVE_DIR)

model.save_pretrained(MODEL_SAVE_DIR)
tokenizer.save_pretrained(MODEL_SAVE_DIR)
print(f"💾 모델 저장 경로: {MODEL_SAVE_DIR}")
```

# 모델 성능 테스트

```python
# ==========================================
# 7. 성능 테스트 (Inference)
# ==========================================
def predict(text):
    model.eval() # 평가 모드로 전환
    
    inputs = tokenizer(
        text, 
        return_tensors='pt', 
        truncation=True, 
        max_length=MAX_LEN, 
        padding='max_length'
    ).to(DEVICE)

    with torch.no_grad():
        outputs = model(**inputs)
        # 확률 계산 (Softmax)
        probs = torch.nn.functional.softmax(outputs.logits, dim=-1)
        pred_idx = torch.argmax(probs, dim=-1).item()
        confidence = probs[0][pred_idx].item()

    result = LABEL_MAP[pred_idx]
    print(f"📝 입력: {text}")
    print(f"🤖 결과: {result} (확신도: {confidence*100:.1f}%)\n")

# --- 테스트 문장 입력 ---
print("--- [AI 감정분석 테스트] ---")
predict("12월 25일 핀크스는 이미 마감이라 어렵습니다.")  # 예상: 불가
predict("김철수님 항공권 발권 완료해드렸습니다.")        # 예상: 확정
predict("현지 호텔에 엑스트라베드 가능한지 확인 중입니다.") # 예상: 보류
predict("주말이라 요금이 좀 비싼데 진행하실까요?")       # 애매한 문장 테스트
predict("사장님 그 날짜는 전멸이라 어렵겠는데요")  # 불가 예상
predict("일단 잡고 30분 뒤에 다시 톡 드릴게요")    # 보류 예상
predict("티켓팅 끝났습니다 파일 확인해주세요")      # 확정 예상

print("--- [심화 검증 테스트: 과연 AI는 똑똑해졌을까?] ---\n")

# 🔴 패턴 1: "일단 잡았다"는 말에 속지 마라! (함정: 확정처럼 보이지만 보류임)
# 아까 틀린 문장: "일단 잡고 30분 뒤에 다시 톡 드릴게요"
predict("좌석은 일단 빼놨는데 입금 확인이 안 되네요")      # 정답: 보류 (2)
predict("가예약 걸어둔 거라 5시까지 결제 없으면 풀립니다") # 정답: 보류 (2)
predict("방은 잡아뒀는데 아직 확정서가 안 나왔어요")       # 정답: 보류 (2)

# 🟠 패턴 2: "부정적 단어(비싸다, 멀다)"에 쫄지 마라! (함정: 거절처럼 보이지만 질문임)
# 아까 틀린 문장: "주말이라 요금이 좀 비싼데 진행하실까요?"
predict("추석이라 요금이 평소 2배인데 괜찮으신가요?")      # 정답: 보류 (2)
predict("원하시는 뷰는 마감이라 가든뷰로 진행할까요?")     # 정답: 보류 (2)

# 🟡 패턴 3: 극단적인 거절 표현 (슬랭/관용구 테스트)
# 아까 맞춘 문장: "사장님 그 날짜는 전멸이라 어렵겠는데요"
predict("그 날짜는 대기도 안 받아줄 정도로 꽉 막혔습니다") # 정답: 불가 (0)
predict("티업이 씨가 말라서 도저히 구할 수가 없네요")      # 정답: 불가 (0)
```
